##4. Limpeza e Preparação dos Dados
A base passa por algumas etapas de preparação para garantir maior consistência durante a análise.

###4.1. Modificações

```python
df = df.drop_duplicates()
```

```python
df["data"] = pd.to_datetime(df["data"])
```

```python
colunas_numericas = [
    "ano",
    "mes",
    "temperatura_media",
    "temperatura_maxima",
    "temperatura_minima",
    "chuva_mm",
    "umidade",
    "velocidade_vento",
    "eventos_extremos"
]
```

```python
for coluna in colunas_numericas:
    df[coluna] = pd.to_numeric(df[coluna], errors="coerce")
```

```python
colunas_texto = [
    "regiao",
    "uf",
    "cidade",
    "nivel_alerta"
]
```

```python
for coluna in colunas_texto:
    df[coluna] = df[coluna].astype(str).str.strip()
```

```python
df["nivel_alerta"] = pd.Categorical(
    df["nivel_alerta"],
    categories=["Baixo", "Médio", "Alto", "Crítico"],
    ordered=True
)
```

```python
df = df.dropna(subset=["data", "regiao", "uf", "cidade"])
```

###Visualização

```python
df.info()
```

##5. Engenharia de Atributos
Cria-se novas variáveis a partir das existentes para facilitar e enriquecer a análise.

###5.1. Variáveis Temporais

```python
df["data"] = pd.to_datetime(df["data"])
df["trimestre"] = df["data"].dt.quarter
df["nome_mes"] = df["data"].dt.month_name()

meses = {
    1: "Janeiro",
    2: "Fevereiro",
    3: "Março",
    4: "Abril",
    5: "Maio",
    6: "Junho",
    7: "Julho",
    8: "Agosto",
    9: "Setembro",
    10: "Outubro",
    11: "Novembro",
    12: "Dezembro"
}
df["nome_mes"] = df["mes"].map(meses)
```

###5.2. Variável: Amplitude Térmica

```python
df["amplitude_termica"] = df["temperatura_maxima"] - df["temperatura_minima"]
```

###5.3. Variável: Classificação de Ocorrência de Evento Extremo

```python
df["teve_evento_extremo"] = np.where(
    df["eventos_extremos"] > 0,
    "Sim",
    "Não"
)
```

##6. Análise Exploratória
Verifica como os dados se comportam, quais regiões se destacam e como os eventos extremos variam ao longo do tempo

###6.1 Estatísticas principais

Como os dados se comportam, quais regiões se destacam e como os eventos extremos variam ao longo do tempo.

```python
df.describe()
```

###6.2 Eventos extremos por região

```python
eventos_regiao = df.groupby("regiao")["eventos_extremos"].sum()
print(eventos_regiao)
```

###6.3 Eventos extremos por ano

```python
eventos_ano = df.groupby("ano")["eventos_extremos"].sum()
print(eventos_ano)
```

###Correlação

```python
colunas = [
    "temperatura_media",
    "chuva_mm",
    "umidade",
    "velocidade_vento",
    "eventos_extremos"
]

correlacao = df[colunas].corr()
print(correlacao.round(2))
```

##7. KPIs
São calculados os principais indicadores do projeto, resumindo os principais aspectos do seu objetivo.

###7.1 Temperatura média

```python
temperatura_media = df["temperatura_media"].mean()
```

###7.2 Chuva média

```python
chuva_media = df["chuva_mm"].mean()
```

###7.3 Total de eventos extremos

```python
total_eventos = df["eventos_extremos"].sum()
```

###7.4 Percentual de registros com eventos

```python
percentual_eventos = (
    df["teve_evento_extremo"].eq("Sim").mean() * 100
)
```

###Visualização

```python
print(f"Temperatura média: {temperatura_media:.2f} °C")
print(f"Chuva média: {chuva_media:.2f} mm")
print(f"Total de eventos extremos: {total_eventos}")
print(f"Registros com eventos extremos: {percentual_eventos:.2f}%")
```

##8. Gráficos

###8.1 Análise Temporal: Eventos Extremos por Ano

```python
plt.figure(figsize=(8, 4))

plt.plot(
    eventos_ano.index,
    eventos_ano.values,
    marker="o"
)

plt.title("Eventos Extremos por Ano")
plt.xlabel("Ano")
plt.ylabel("Quantidade de Eventos")
plt.grid(True)
plt.show()
```

###8.2 Eventos extremos por região

```python
plt.figure(figsize=(8, 4))

sns.barplot(
    x=eventos_regiao.index,
    y=eventos_regiao.values
)

plt.title("Eventos Extremos por Região")
plt.xlabel("Região")
plt.ylabel("Quantidade de Eventos")
plt.xticks(rotation=30)
plt.show()
```

###Correlação

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    correlacao,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlação entre Variáveis")
plt.show()
```
